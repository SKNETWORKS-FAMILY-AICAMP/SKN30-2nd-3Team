```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np
```

```python
# 호텔 예약 데이터에서 필요한 날짜 추출
# 날씨 데이터는 현지 날짜가 유의미한 날씨이므로 실제 호텔 도착 날을 기준으로 함
df_hotel = pd.read_csv('../data/preprocessed/hotel_bookings.csv')

date_min = df_hotel['arrival_date'].min()
date_max = df_hotel['arrival_date'].max()

print(f'도착 날짜: {date_min} ~ {date_max}')
```

```python
# 날씨 데이터 불러오기
df = pd.read_csv('../data/raw/weather.csv')
df.head()
```

```python
# 데이터 정보
df.info()
```

```python
# 날짜 데이터 타입 변경
df['date'] = pd.to_datetime(df['date'].str[:10], format='%Y-%m-%d')
df['date']
```

```python
df.describe()
```

```python
# 결측치 확인
df.isna().sum()
```

```python
# 결측치 처리
# tavg: 평균 기온 -----> tmin, tmax 평균값
# 평균 기온, 강수량 외 결측치가 포함된 열 삭제
missing_cols_drop = ['snow', 'wdir', 'wspd', 'wpgt', 'pres', 'tsun']
missing_cols_fill = ['tavg']

df.drop(missing_cols_drop, axis=1, inplace=True)
df[missing_cols_fill[0]] = df[missing_cols_fill[0]].fillna(df[['tmin', 'tmax']].mean(axis=1))

df.isna().sum()
```

```python
# 파일 저장
df.to_csv('../data/preprocessed/weather.csv', index=False)
```
